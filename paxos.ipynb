```python
class Acceptor:

    def __init__(self, id):
        self.id = id
        self.promised_id = -1
        self.accepted_id = -1
        self.accepted_value = None

    def prepare(self, proposal_id):

        if proposal_id > self.promised_id:
            self.promised_id = proposal_id
            print(f"Acceptor {self.id}: Promise for Proposal {proposal_id}")
            return True

        print(f"Acceptor {self.id}: Reject Proposal {proposal_id}")
        return False

    def accept(self, proposal_id, value):

        if proposal_id >= self.promised_id:

            self.promised_id = proposal_id
            self.accepted_id = proposal_id
            self.accepted_value = value

            print(f"Acceptor {self.id}: Accepted '{value}'")
            return True

        return False


class Proposer:

    def __init__(self, acceptors):
        self.acceptors = acceptors

    def propose(self, proposal_id, value):

        print("\n----- PREPARE PHASE -----")

        promises = 0

        for acceptor in self.acceptors:
            if acceptor.prepare(proposal_id):
                promises += 1

        majority = len(self.acceptors)//2 + 1

        if promises < majority:
            print("\nConsensus Failed (No Majority)")
            return

        print("\nMajority Promises Received")

        print("\n----- ACCEPT PHASE -----")

        accepts = 0

        for acceptor in self.acceptors:
            if acceptor.accept(proposal_id, value):
                accepts += 1

        if accepts >= majority:
            print("\nConsensus Achieved")
            print("Chosen Value:", value)
        else:
            print("\nConsensus Failed")


# --------------------------

acceptors = [
    Acceptor(1),
    Acceptor(2),
    Acceptor(3),
    Acceptor(4),
    Acceptor(5)
]

proposer = Proposer(acceptors)

proposer.propose(1, "BLOCK_100")
```
