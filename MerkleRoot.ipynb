```python
import hashlib

# Function to compute SHA-256 hash
def sha256(data):
    return hashlib.sha256(data.encode()).hexdigest()

# Function to compute Merkle Root
def merkle_root(transactions):
    # Step 1: Hash all transactions
    hashes = [sha256(tx) for tx in transactions]

    print("Initial Transaction Hashes:")
    for i, h in enumerate(hashes, start=1):
        print(f"H{i}: {h}")

    level = 1

    # Step 2: Build Merkle Tree
    while len(hashes) > 1:

        print(f"\nLevel {level}")

        # Duplicate last hash if odd number of hashes
        if len(hashes) % 2 != 0:
            hashes.append(hashes[-1])

        new_hashes = []

        for i in range(0, len(hashes), 2):
            combined = hashes[i] + hashes[i + 1]
            new_hash = hashlib.sha256(combined.encode()).hexdigest()

            print(f"Hash({i+1},{i+2}) = {new_hash}")

            new_hashes.append(new_hash)

        hashes = new_hashes
        level += 1

    return hashes[0]


# ---------------- Main Program ----------------

transactions = [
    "Alice pays Bob 5 BTC",
    "Bob pays Charlie 2 BTC",
    "Charlie pays David 1 BTC",
    "David pays Eve 3 BTC"
]

root = merkle_root(transactions)

print("\nMerkle Root:")
print(root)
```
